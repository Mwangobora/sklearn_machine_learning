```python
import pandas as pd
heart_disease = pd.read_csv("heart.csv")
heart_disease
```

```python
#create X (feature matrxix)
x = heart_disease.drop("target", axis=1)

#create Y (label)
y = heart_disease["target"]
```

```python
#choose the tight model and hyperparamaters
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier(n_estimators=100)
#we'll keep the default hyperparameters
import sklearn
sklearn.show_versions()
```

```python
#fit the model to the training data

from sklearn.model_selection import train_test_split    
x_train,x_test, y_train,y_test = train_test_split(x,y, test_size=0.2)
x_train
```

```python
clf.fit(x_train, y_train)
```

```python
#make a prediction
import numpy as np
```

```python
y_preds = clf.predict(x_test)
y_preds
```

```python
y_test
```

```python
#evaluate the model on training the dataset
clf.score(x_train, y_train)
```

```python
clf.score(x_test,y_test)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
print(classification_report(y_test, y_preds))
```

```python
confusion_matrix(y_test, y_preds)
```

```python
accuracy_score(y_test, y_preds)
```

```python
#improve the model
#try different types of estimators

np.random.seed(42)

for i in range(10,100, 10):
    print(f"try the model with {i} estimators....")
    clf = RandomForestClassifier(n_estimators=i).fit(x_train,y_train)
    print(f"MOdal accuracy on test set: {clf.score(x_test, y_test)*100:.2f}%")
    print()
```

```python
#save the model and load it

import pickle
pickle.dump(clf, open("random_forest1.pkl", "wb"))
```

```python
#laod the model

loaded_model = pickle.load(open("random_forest1.pkl","rb"))
loaded_model.score(x_test,y_test)
```
